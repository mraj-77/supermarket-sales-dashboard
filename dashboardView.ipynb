```python
total_sales = df["Sales"].sum()

print("Total Sales:", total_sales)
```

```python
total_orders = df["Invoice ID"].nunique()

print("Total Orders:", total_orders)
```

```python
average_rating = df["Rating"].mean()

print("Average Rating:", round(average_rating, 2))
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
print("================================")
print("       SALES DASHBOARD")
print("================================")

print("Total Sales       :", round(df["Sales"].sum(), 2))
print("Total Orders      :", df["Invoice ID"].nunique())
print("Average Rating    :", round(df["Rating"].mean(), 2))
print("Quantity Sold     :", df["Quantity"].sum())
```

```python
sales_product = df.groupby("Product line")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))

sales_product.plot(kind="bar")

plt.title("Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales_city = df.groupby("City")["Sales"].sum()

plt.figure(figsize=(7, 5))

sales_city.plot(kind="bar")

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
sales_branch = df.groupby("Branch")["Sales"].sum()

plt.figure(figsize=(7, 5))

sales_branch.plot(kind="bar")

plt.title("Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
payment_count = df["Payment"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    payment_count,
    labels=payment_count.index,
    autopct="%1.1f%%"
)

plt.title("Payment Method Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Rating"],
    df["Sales"]
)

plt.title("Rating vs Sales")
plt.xlabel("Rating")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

```python
sales_branch = df.groupby("Gender")["Sales"].sum()

plt.figure(figsize=(7, 5))

sales_branch.plot(kind="bar")

plt.title("Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(20, 20))
axes[0,0].set_xticklabels(sales_product.index, rotation=45, ha='right')
axes[0,0].bar(sales_product.index, sales_product.values)
axes[0,0].set_title("Sales by Product")
axes[0,1].bar(sales_city.index, sales_city.values)
axes[0,1].set_title("Sales by City")
axes[1,0].bar(sales_branch.index, sales_branch.values)
axes[1,0].set_title("Sales by Branch")
axes[1,1].pie(
    payment_count,
    labels=payment_count.index, 
)
axes[1,1].set_title("Payment Method Distribution")
axes[0,2].scatter(df["Rating"], df["Sales"])
axes[0,2].set_title("Rating vs Sales")
axes[0,2].set_xlabel("Rating")
axes[0,2].set_ylabel("Sales")
axes[1,2].bar(sales_branch.index, sales_branch.values)
axes[1,2].set_title("Sales by Gender")
```
